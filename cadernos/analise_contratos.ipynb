# Análise de Contratos Imobiliários

Projeto de análise exploratória de contratos imobiliários utilizando Python, Pandas e Matplotlib.

**Objetivo:** identificar padrões de aluguel, distribuição por tipo de imóvel e cidade, receitas de administração e indicadores de inadimplência.

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

## 1. Carregamento da base

```python
df = pd.read_csv('../dados/contratos_imobiliarios.csv')
df.head()
```

## 2. Visão geral dos dados

```python
print(f'Quantidade de contratos: {len(df)}')
print(f'Quantidade de colunas: {df.shape[1]}')
df.info()
```

```python
df.describe(include='all').T
```

## 3. Tratamento inicial

```python
df['inicio_contrato'] = pd.to_datetime(df['inicio_contrato'])
df['fim_contrato'] = pd.to_datetime(df['fim_contrato'])
df['inadimplente'] = df['inadimplente'].astype(int)
df.isna().sum()
```

## 4. Valor médio do aluguel por tipo de imóvel

```python
media_tipo = df.groupby('tipo_imovel')['valor_aluguel'].mean().sort_values(ascending=False)
media_tipo
```

```python
media_tipo.plot(kind='bar', figsize=(10, 5), title='Aluguel médio por tipo de imóvel')
plt.ylabel('Valor médio do aluguel (R$)')
plt.xlabel('Tipo de imóvel')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 5. Distribuição dos contratos por cidade

```python
contratos_cidade = df['cidade'].value_counts()
contratos_cidade
```

```python
contratos_cidade.plot(kind='bar', figsize=(10, 5), title='Quantidade de contratos por cidade')
plt.ylabel('Quantidade de contratos')
plt.xlabel('Cidade')
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

## 6. Receita estimada de administração

```python
receita_total = df['taxa_administracao'].sum()
receita_media = df['taxa_administracao'].mean()
print(f'Receita mensal estimada de administração: R$ {receita_total:,.2f}')
print(f'Receita média por contrato: R$ {receita_media:,.2f}')
```

## 7. Indicador de inadimplência

```python
taxa_inadimplencia = df['inadimplente'].mean() * 100
print(f'Taxa de inadimplência da carteira: {taxa_inadimplencia:.2f}%')
```

```python
inadimplencia_garantia = (df.groupby('garantia')['inadimplente'].mean() * 100).sort_values(ascending=False)
inadimplencia_garantia
```

## 8. Conclusões de negócio

- Comparar quais tipos de imóveis apresentam maior aluguel médio.
- Identificar as cidades com maior concentração da carteira.
- Avaliar a receita mensal estimada de administração.
- Monitorar a taxa de inadimplência e sua relação com o tipo de garantia.
- Utilizar os resultados para apoiar decisões comerciais e de gestão da carteira.
